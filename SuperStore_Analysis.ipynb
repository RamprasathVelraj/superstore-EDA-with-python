```python
import seaborn as sns
import matplotlib
import matplotlib.pyplot as plt
```

```python
# To Visualize Categories percentage Contribution
plt.title('Category Percentage Distribution')
df['Category'].value_counts().plot.pie(autopct='%1.1f%%',figsize=(7,5))
plt.show()
```

```python
plt.figure(figsize=(6, 4))
plt.bar(df['Category'], df['Sales'], color='skyblue')

plt.title('Sales based on Category')
plt.xlabel('Category')
plt.ylabel('Sales')
plt.show()
```

```python
plt.figure(figsize=(6,4))
df_sort = df.sort_values(by='Sales',ascending= False)
plt.barh(df_sort['Sub-Category'],df_sort['Sales'],color = 'orange')
plt.gca().invert_yaxis()
plt.title('Sales Based on Sub-Category')
plt.xlabel('Sub-Category')
plt.ylabel('Sales')
plt.show()
```

```python
sns.violinplot(x='Region', y='Sales', data=df)
plt.show()
```

```python
grouped = df.groupby('Region', as_index=False)['Quantity'].sum()
sns.barplot(x='Region', y='Quantity', data=grouped)
plt.title('Total Quantity Sold per Region')
plt.xlabel('Region')
plt.ylabel('Quantity Sold')
plt.tight_layout()
plt.show()
```

```python
graph= df.corr(method = 'pearson', numeric_only = True)
sns.heatmap(graph, cmap = "Blues")
plt.title("Rep")
```

```python
grouped = df.groupby('Region')['Quantity'].sum()
labels = grouped.index
sizes = grouped.values

# Plot donut chart
fig, ax = plt.subplots()
ax.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=90, 
       wedgeprops={'width': 0.4})  # 'width' creates the donut hole
ax.axis('equal')  # Ensures it's a circle
plt.title('Quantity Sold by Region')
plt.tight_layout()
plt.show()
```
